```python
import pandas as pd
import numpy as np

df = pd.read_excel('../data/Gasverbräuche Bensheim_Etagenheizungen.xlsx')
```

# Process raw data of Gas consumption in Bensheim

```python
df.drop(columns=['Gemeindename', 'Hausnummer', 'Straßenname'], inplace=True)
```

```python
years_array = np.arange(2012, 2024)
agg_functions = {}
for year in years_array:
    agg_functions[year] = 'sum'

# Group by 'Straßenname' and 'Hausnummer' and apply aggregation functions
grouped_df = df.groupby(['Geographische Höhe (ü. NHN) (m)', 'X B Position ETRS89 Zone 32', 'Y B Position ETRS89 Zone 32']).sum().reset_index()
grouped_df = grouped_df.drop(columns=['Geographische Höhe (ü. NHN) (m)'])

grouped_df.index.name = 'id'
```

```python
# Rename columns
grouped_df.rename(columns={
    'X B Position ETRS89 Zone 32': 'xcoord',
    'Y B Position ETRS89 Zone 32': 'ycoord'
}, inplace=True)

# Convert coordinates from strings to floats
grouped_df['xcoord'] = grouped_df['xcoord'].str.replace(' m', '').astype(float)
grouped_df['ycoord'] = grouped_df['ycoord'].str.replace(' m', '').astype(float)
```

```python
grouped_df.to_excel('../data/Gasverbräuche Bensheim_Etagenheizungen_grouped.xlsx')
grouped_df.to_csv('../data/Gasverbräuche Bensheim_Etagenheizungen_grouped.csv')
```
